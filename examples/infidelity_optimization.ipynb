# Infidelity minimisation

This notebook shows how to fit a variational state to a target state by minimising the infidelity, with `jVMC.objective_function.Infidelity`. The target is a reference state $|\chi\rangle$ transformed by an operator $\hat O$, and the infidelity is

$$
\mathcal I = 1 - \frac{\langle\psi_\theta|\hat O|\chi\rangle\langle\chi|\hat O^\dagger|\psi_\theta\rangle}{\langle\psi_\theta|\psi_\theta\rangle\langle\chi|\chi\rangle}.
$$

Both $|\psi_\theta\rangle$ and $|\chi\rangle$ are sampled, so $|\chi\rangle$ can itself be a neural quantum state. This is, for example, the building block of time evolution with projected methods, where $\hat O$ is a short-time propagator. Dense one- and two-site matrices are turned into operators with `jVMC.util.util.matrix_to_jvmc_operator`.

The example is structured as follows:

1. Build an exact reference ground state for a small transverse-field Ising chain.
2. Convert a dense two-site unitary into a `jVMC` operator with `matrix_to_jvmc_operator`.
3. Verify the operator conversion with an exact infidelity test.
4. Optimise a variational state by minimising the infidelity.
5. Compare the estimator against the exact overlap at the end.

```python
import matplotlib.pyplot as plt
import numpy as np
import jax.numpy as jnp

import jVMC
import jVMC.nets as nets
import jVMC.operator.discrete as op
from jVMC.util.util import matrix_to_jvmc_operator
from jVMC.vqs import NQS
```

## Build the reference state

We use a periodic transverse-field Ising chain of length $L = 6$,

$$
H = -\frac{g}{2}\sum_{i} \sigma^x_i + \frac{J}{2}\sum_{i} \sigma^z_i \sigma^z_{i+1},
$$

with an antiferromagnetic coupling $J = 1$ and $g = 0.5$, and obtain its exact ground state $|\chi\rangle$ from dense diagonalisation. The helper `embed_local_operator` is only used to build the dense reference Hamiltonian and to embed a two-site matrix into the full Hilbert space for later validation. Site 0 is the leftmost factor of the Kronecker products, which matches the ordering of the basis states in `jVMC`.

```python
sx = np.array([[0.0, 1.0], [1.0, 0.0]], dtype=complex)
sz = np.array([[1.0, 0.0], [0.0, -1.0]], dtype=complex)


def embed_local_operator(local_matrix, site, system_size, nsites=1):
    if site == 0:
        return np.kron(local_matrix, np.identity(2 ** (system_size - nsites), dtype=complex))
    return np.kron(
        np.identity(2 ** site, dtype=complex),
        np.kron(local_matrix, np.identity(2 ** (system_size - site - nsites), dtype=complex)),
    )


def create_tfim_hamiltonian(g, J, system_size):
    transverse = embed_local_operator(sx, 0, system_size)
    for site in range(1, system_size):
        transverse += embed_local_operator(sx, site, system_size)

    interaction = (
        embed_local_operator(sz, 0, system_size) @ embed_local_operator(sz, 1, system_size)
        + embed_local_operator(sz, system_size - 1, system_size) @ embed_local_operator(sz, 0, system_size)
    )
    for site in range(1, system_size - 1):
        interaction += embed_local_operator(sz, site, system_size) @ embed_local_operator(sz, site + 1, system_size)

    return -0.5 * (g * transverse - J * interaction)


L = 6
J = 1.0
g = 0.5

eigenvalues, eigenvectors = np.linalg.eigh(create_tfim_hamiltonian(g, J, L))
# The off-diagonal elements of H are non-positive, so the ground state can be chosen positive:
# the absolute value only fixes the arbitrary global sign
ground_state = np.abs(eigenvectors[:, 0])
ground_state = ground_state / np.linalg.norm(ground_state)

print(f"Ground-state energy: {eigenvalues[0]:.8f}")
print(f"Hilbert-space dimension: {2 ** L}")
```

## Convert a dense two-site matrix into an operator

Next we create a random two-site unitary $U$ and convert it with `matrix_to_jvmc_operator`. The matrix is expanded in products of Pauli matrices on the given sites. With `return_conjugate=True` it also returns the operator of the adjoint $U^\dagger$, which `Infidelity` needs together with $U$. We also embed the same dense matrix into the full Hilbert space, so that later comparisons can be performed directly in the computational basis.

The same helper also supports single-site operators: passing a `2 x 2` matrix together with `sites=(i,)` returns the corresponding one-site operator acting on site `i`.

```python
def haar_unitary(n, rng):
    ginibre = (rng.normal(size=(n, n)) + 1j * rng.normal(size=(n, n))) / np.sqrt(2.0)
    q, r = np.linalg.qr(ginibre)
    phases = np.diag(r) / np.abs(np.diag(r))
    return q * phases


def random_unitary(n=4, eps=0.2, seed=1234):
    rng = np.random.default_rng(seed)
    basis_change = haar_unitary(n, rng)
    generator_eigs = rng.uniform(-eps, eps, size=n)
    hermitian = basis_change @ np.diag(generator_eigs) @ basis_change.conj().T
    w, v = np.linalg.eigh(hermitian)
    return v @ np.diag(np.exp(-1j * w)) @ v.conj().T


sites = (0, 1)
unitary_local = random_unitary(n=4, eps=1.0, seed=452)
unitary_jvmc, conjugate_unitary_jvmc = matrix_to_jvmc_operator(
    unitary_local,
    sites,
    return_conjugate=True,
)
unitary_dense = embed_local_operator(unitary_local, site=0, system_size=L, nsites=2)

print(f"Local operator is unitary: {np.allclose(unitary_local.conj().T @ unitary_local, np.eye(4))}")
```

## Test 1: verify `matrix_to_jvmc_operator` with exact sampling

This first test checks that the converted operator acts like the original dense matrix. We create

- a reference state `chi`, holding the Ising ground state $|\chi\rangle$,
- a state `chi_transformed`, holding $U|\chi\rangle$ computed with the dense matrix,
- an `Infidelity` objective with reference `chi` and the converted operators $U$, $U^\dagger$.

`FullConfigurationState` is a network with one complex parameter per basis state, used here to load exact state vectors; the flat parameter array holds the real parts followed by the imaginary parts. The `ExactSampler` sums over all basis states, so the result is exact. If the conversion is correct, the infidelity of `chi_transformed` is zero up to rounding errors.

```python
reference_model = nets.FullConfigurationState(L=L, d=2)
chi = NQS(reference_model, L, batch_size=2 ** L)
chi.parameters = jnp.concatenate([jnp.real(ground_state), jnp.imag(ground_state)])
chi_sampler = jVMC.sampler.ExactSampler(chi)

transformed_state = unitary_dense @ ground_state
transformed_state /= np.linalg.norm(transformed_state)

chi_transformed = NQS(reference_model, L, batch_size=2 ** L)
chi_transformed.parameters = jnp.concatenate([jnp.real(transformed_state), jnp.imag(transformed_state)])
chi_transformed_sampler = jVMC.sampler.ExactSampler(chi_transformed)

loss_function = jVMC.objective_function.Infidelity(
    reference_sampler=chi_sampler,
    operator=unitary_jvmc, 
    conjugated_operator=conjugate_unitary_jvmc
)

infidelity = loss_function(chi_transformed_sampler)
infidelity_mean = infidelity.mean.real.item()

print(f"Infidelity after dense-to-jVMC conversion: {infidelity_mean}")
print(f"Conversion test passed: {np.allclose(infidelity_mean, 0.0)}")
```

## Optimise a trial state

We now minimise the infidelity of a variational complex RBM with respect to the transformed target state $U|\chi\rangle$. The reference `chi` itself remains unchanged; the operators supplied to `Infidelity` define the transformation.

This time both states are sampled with Monte Carlo, with independent samplers (different seeds). `control_variate="adaptive"` reduces the variance of the infidelity estimate without changing its gradient. The optimiser is stochastic reconfiguration, as for a ground state search: `ground_state_search` minimises any objective function, and the progress bar shows the current infidelity estimate instead of an energy.

```python
num_hidden = 2**6
num_iterations = 200
num_samples = 2**9
num_chains = 2**7
learning_rate = 1e-1

trial_model = nets.CpxRBM(num_hidden=num_hidden, bias=True)
psi = NQS(trial_model, L, batch_size=num_samples)

proposer = jVMC.propose.SpinFlip()
chi_sampler = jVMC.sampler.MCSampler(chi, proposer, 456, num_chains, num_samples)
psi_sampler = jVMC.sampler.MCSampler(psi, proposer, 123, num_chains, num_samples)

loss_function = jVMC.objective_function.Infidelity(
    reference_sampler=chi_sampler,
    operator=unitary_jvmc, 
    conjugated_operator=conjugate_unitary_jvmc,
    control_variate="adaptive"
)
stepper = jVMC.stepper.Euler(learning_rate)
opt = jVMC.optimizer.SR(psi_sampler, psi)

out = opt.ground_state_search(num_iterations, loss_function, stepper)
```

## Test 2: compare the final estimator to the exact overlap

After optimisation we validate the result in two ways:

- the infidelity estimate should decrease overall during the optimisation,
- the final infidelity, computed with exact sampling, should agree with the one computed directly from the dense state vectors, $1 - |\langle U\chi|\psi_\theta\rangle|^2$ for normalised states.

This provides an end-to-end check of both `matrix_to_jvmc_operator` and `Infidelity` in a realistic workflow.

```python
infidelity_history = np.array(out['energy']['mean'])
fig, ax = plt.subplots(figsize=(6, 4), dpi=120)
ax.plot(np.arange(num_iterations), np.abs(infidelity_history))
ax.set_yscale("log")
ax.set_xlabel("Iteration")
ax.set_ylabel("Estimated infidelity")
ax.set_title("Infidelity minimisation")
plt.show()

psi_exact_sampler = jVMC.sampler.ExactSampler(psi)
chi_exact_sampler = jVMC.sampler.ExactSampler(chi)

exact_validation = jVMC.objective_function.Infidelity(
    chi_exact_sampler, unitary_jvmc, conjugate_unitary_jvmc 
)

infidelity = exact_validation(psi_exact_sampler)

estimated_final_infidelity = infidelity.mean.real.item()

psi_state = np.exp(np.asarray(psi_exact_sampler.log_psi).reshape(-1))
psi_state = psi_state / np.linalg.norm(psi_state)
reference_state = transformed_state / np.linalg.norm(transformed_state)
exact_overlap_infidelity = float(np.real(1.0 - np.abs(np.vdot(reference_state, psi_state)) ** 2))

print(f"Final exact-sampling estimator: {estimated_final_infidelity:.5e}")
print(f"Final exact overlap value:      {exact_overlap_infidelity:.5e}")
print(f"Estimator matches overlap:      {np.allclose(estimated_final_infidelity, exact_overlap_infidelity)}")
```
